```
import cv2 
import os 
import json
import numpy as np 
import matplotlib.pyplot as plt
from tqdm import tqdm
from imutils import perspective
```

```
CODE_DATASET_PATH = "/root/work/LogicMate_v1/datasets/Code-Charaters-Detection-1"
```

```
CODE_DATASET_TRAIN_FILE = os.path.join(CODE_DATASET_PATH, 'train')
```

```
CODE_DATASET_VAL_FILE = os.path.join(CODE_DATASET_PATH, 'val')
```

```
CODE_DATASET_TRAIN_ANNOTATION = os.path.join(CODE_DATASET_TRAIN_FILE, 'train_annotation.txt')
```

```
CODE_DATASET_VAL_ANNOTATION = os.path.join(CODE_DATASET_VAL_FILE, 'valid_annotation.txt')
```

```
import os
import json
from tqdm import tqdm
import cv2

annotations_file = CODE_DATASET_TRAIN_ANNOTATION
dataset_path = CODE_DATASET_TRAIN_FILE # Ruta completa
output_file = "train.txt"

count = 0

if not os.path.exists(annotations_file):
    print(f"Error: No se encontró {annotations_file}")
    exit()

with open(annotations_file, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("\n Verificando líneas en train_annotation.txt:")
for line in lines[:5]:  # Mostrar solo las primeras 5 líneas
    print(f"  {line.strip()}")

with open(output_file, "w", encoding="utf-8") as ob:
    for line in tqdm(lines):
        parts = line.strip().split(" ", 1)  # Separar imagen y texto
        if len(parts) != 2:
            print(f"Error en línea: {line.strip()}")  # Detectar líneas incorrectas
            continue

        img_name, text = parts
        img_path = os.path.join(dataset_path, img_name)

        if not os.path.exists(img_path):
            print(f"Imagen no encontrada: {img_path}")
            continue  # Omitir imágenes que no existen

        # Obtener dimensiones de la imagen
        image = cv2.imread(img_path)
        if image is None:
            print(f"No se pudo cargar la imagen: {img_path}")
            continue

        h, w, _ = image.shape  # Altura y ancho

        # Generar bounding box en toda la imagen (asumiendo que el texto está centrado)
        bbox = [[0, 0], [w, 0], [w, h], [0, h]]

        # Crear formato JSON para PaddleOCR
        annotation_json = [{"points": bbox, "transcription": text}]

        # Guardar en train.txt
        ob.write(img_path + "\t" + json.dumps(annotation_json) + "\n")
        count += 1

print(f"\nTotal de anotaciones convertidas: {count}")
print("\nVerificando contenido de train.txt:")

with open(output_file, "r", encoding="utf-8") as f:
    for line in f.readlines()[:5]:  # Mostrar solo las primeras 5 líneas
        print(f"  {line.strip()}")
```
